Histograma

```python
m=2**31-1
a=1103515245
b=12345
x=749321

numeros = []

for i in range(10**4):
    x = (a*x + b) % m
    numeros.append(x)

print(numeros)
import matplotlib.pyplot as plt
plt.hist(numeros, bins=100)
plt.show()
```

Histograma de 2-5

```python
import numpy as np
numeros=np.array(numeros)
#distribucion uniforme de numeros del 2-5)
numeros_uniforme = 2 + 6 * numeros / m
plt.hist(numeros_uniforme, bins=100)
plt.show()
```
