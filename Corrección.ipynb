## Ejercicio 1

**¿Cuál fue el error?**

Solo estaba guardando la ultima volatilidad entonces no estaba tomando todas las ventanas de 20 datos

**¿Cuál es la corrección?**

Guardar la volatilidad de cada ventana dentro del ciclo

**¿Por qué se cometió el error?**

Porque hice el ciclo pero no me fije que al final solo estaba guardando el ultimo resultado

**¿Cómo se puede evitar este error en el futuro?**

Revisar cuantos resultados me salen y comprobar que si se este guardando un resultado por cada ventana

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_excel('ex1_01.xlsx')

df['Rendimiento_A']=np.log(df['Ruta A']/df['Ruta A'].shift(1))
df['Rendimiento_B']=np.log(df['Ruta B']/df['Ruta B'].shift(1))

rendimiento_A=df['Rendimiento_A'].dropna()
rendimiento_B=df['Rendimiento_B'].dropna()

n=20

vol_1=rendimiento_A.rolling(n).std().dropna()
vol_2=rendimiento_B.rolling(n).std().dropna()

vol_1_annual=vol_1*np.sqrt(252)
vol_2_annual=vol_2*np.sqrt(252)

print('Máxima volatilidad anual de A:',np.max(vol_1_annual))
print('Máxima volatilidad anual de B:',np.max(vol_2_annual))
```

El contrato se activa en la ruta que no pase el 20% de volatilidad anualizada

## Ejercicio 2

**¿Cuál fue el error?**

Solo hice una trayectoria aunque M era 10000 y tampoco agregue los cambios de tasa volatilidad y el anuncio

**¿Cuál es la corrección?**

Hacer las 10000 simulaciones dia por dia y cambiar la tasa y la volatilidad en los dias que pide el problema tambien agregar el efecto positivo del anuncio

**¿Por qué se cometió el error?**

Porque use el GBM normal y no lo cambie con las condiciones que daba el problema

**¿Cómo se puede evitar este error en el futuro?**

Antes de hacer el codigo separar los cambios por dia y al final revisar que si haya usado todas las condiciones que me dieron

Para el anuncio del dia 120 como no dice cuanto aumenta voy a tomar una semana como 5 dias y durante esos dias hacer positivos los valores de Z

```python
S0=100
r=.05
sigma=.20
T=1
n=252
M=10000

dt=T/n

np.random.seed(32)

precios=np.zeros((M,n+1))
precios[:,0]=S0

for t in range(1,n+1):
    if t>=30:
        r=.08
    else:
        r=.05

    if 50<=t<70:
        sigma=.40
    else:
        sigma=.20

    Z=np.random.normal(0,1,size=M)

    if 120<=t<125:
        Z=np.abs(Z)

    precios[:,t]=precios[:,t-1]*np.exp((r-.5*sigma**2)*dt+sigma*np.sqrt(dt)*Z)

precios.shape
```

La tasa cambia a 8% desde el dia 30

La volatilidad cambia a 40% durante 20 dias y despues vuelve a 20%

Del dia 120 al 124 hago positivos los valores de Z por el anuncio

```python
plt.figure(figsize=(10,6))

for i in range(20):
    plt.plot(precios[i])

plt.axvline(30,linestyle='--')
plt.axvline(50,linestyle='--')
plt.axvline(70,linestyle='--')
plt.axvline(120,linestyle='--')
plt.axvline(125,linestyle='--')

plt.xlabel('Dias')
plt.ylabel('Precio')
plt.title('20 trayectorias')
plt.show()
```
