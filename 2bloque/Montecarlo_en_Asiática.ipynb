```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
```

```python
S0 = 100
K = 100
sigma = 0.20
r = 0.05
T = 1
q = 0

b = r - q
```

```python
sigma_A = sigma / np.sqrt(3)
b_A = 1/2 * (b - (sigma_A**2 / 2))
d1=np.log(S0 / K) + (b_A + sigma_A**2 / 2) * T / (sigma_A * np.sqrt(T))
d2=d1 - sigma_A * np.sqrt(T)

print ("sigma_A: ", sigma_A)
print ("b_A: ", b_A)
print ("d1: ", d1)
print ("d2: ", d2)
```

```python
c=S0 *np.exp((b_A-r)*T)* norm.cdf(d1)-K*np.exp(-r*T)*norm.cdf(d2)
c
```
