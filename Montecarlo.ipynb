Act 1 calcular el precio final

```python
S0=100
K=110
r=.05
V=0.20
T=1
M=10000 # cantidad de simulaciones
precios=[]
import numpy as np
import matplotlib.pyplot as plt
Z= np.random.normal(0,1, size=M)
ST= S0*np.exp((r-0.5*V**2)*T+V*np.sqrt(T)*Z)
ST




#
```

```python
np.random.seed(32)

precios = []

for i in range(M):
    Z = np.random.normal(0, 1)

    ST = S0 * np.exp((r - (1/2) * V**2) * T + V * np.sqrt(T) * Z)

    precios.append(ST)

plt.hist(precios, bins=1000)
plt.show()
```

Ejercicio: compre un call con precio de ejercicio de 110 , universo en donde el precio de la accion es menor a 110, (igualarlos a 0)
arriba de 110 es s(T)-k
y abajo de 110 igualarlo a 0

```python
np.random.seed(42)
Call=[]
for i in precios:

    if i > K:
        Call.append((i - K)*np.exp(-r*T)) # esto lo cambio a valor presente , 
        #porq sin el *np.exp(-r*T) estaba en futuro y usando *np.exp(-r*T) se paso a valor presente

    else:
        Call.append(0)

Call
plt.hist(Call, bins=50)
plt.show()
```

calcular dos valores distintos:
con la paqueteria scipy.stats

```python
from scipy.stats import norm
```

```python
d1 = (np.log(S0 / K) + (r + (1/2) * V**2) * T) / (V * np.sqrt(T))
d2 = d1 - V * np.sqrt(T)
n1 = norm.cdf(d1)
n2 = norm.cdf(d2)
C = S0 * n1 - K * np.exp(-r * T) * n2
print(C)
```
